```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset, Subset
import numpy as np
from tqdm.auto import tqdm
from PIL import Image
from torchvision import transforms
from torchvision.utils import save_image
import matplotlib.pyplot as plt

# --- 1. 설정 (Settings) ---
DEVICE = torch.device("cuda:2" if torch.cuda.is_available() else "cpu")
GPU_IDS = [2, 1, 3] # 사용할 GPU ID 목록

# --- 경로 설정 ---
DATA_DIR = '/home/nas/data/YMG/datas/celeba_hq_256/'
AE_CHECKPOINT_DIR = '/home/nas/data/YMG/superdeep_ae/checkpoints/' # 학습된 VAE-GAN 가중치 경로
SAVE_DIR = '/home/nas/data/YMG/superdeep_ae/보간넷실험03_GAN과적합/' # 이번 실험 결과 저장 경로
os.makedirs(SAVE_DIR, exist_ok=True)
INTERP_NET_SAVE_PATH = os.path.join(SAVE_DIR, 'interpolation_net_gan_overfit.pth')

# --- 하이퍼파라미터 ---
LATENT_DIM = 256
LEARNING_RATE = 1e-5 # GAN 학습은 불안정할 수 있으므로 학습률을 약간 낮춤
BATCH_SIZE = 100
NUM_EPOCHS = 100 # 과적합을 위해 충분히 많이 학습
REPROJ_WEIGHT = 0.5 # 기하학적 일관성 손실 가중치 (lambda)
ADV_WEIGHT = 1.0    # 시각적 현실성 손실 가중치 (gamma)

# 테스트에 고정적으로 사용할 이미지 쌍 인덱스
FIXED_PAIRS_INDICES = [
    (9141, 21742), (1857, 17388), (9926, 28731),
    (7732, 1726), (27763, 26175)
]
NUM_POINTS_PER_PAIR = 100

# --- 2. 모델 및 헬퍼 함수 정의 (이전과 동일) ---
class ResBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.block = nn.Sequential(nn.Conv2d(in_channels, out_channels, 3, 1, 1), nn.BatchNorm2d(out_channels), nn.SiLU(), nn.Conv2d(out_channels, out_channels, 3, 1, 1), nn.BatchNorm2d(out_channels))
        self.shortcut = nn.Conv2d(in_channels, out_channels, 1) if in_channels != out_channels else nn.Identity()
        self.silu = nn.SiLU()
    def forward(self, x): return self.silu(self.block(x) + self.shortcut(x))

class EncoderSuperDeep(nn.Module):
    def __init__(self, in_channels=3, base_channels=128, latent_channels=1):
        super().__init__()
        self.encoder = nn.Sequential(nn.Conv2d(in_channels, base_channels, 3, 1, 1), ResBlock(base_channels, base_channels), nn.Conv2d(base_channels, base_channels*2, 3, 2, 1), ResBlock(base_channels*2, base_channels*2), nn.Conv2d(base_channels*2, base_channels*4, 3, 2, 1), ResBlock(base_channels*4, base_channels*4), nn.Conv2d(base_channels*4, base_channels*8, 3, 2, 1), ResBlock(base_channels*8, base_channels*8), nn.Conv2d(base_channels*8, base_channels*16, 3, 2, 1), ResBlock(base_channels*16, base_channels*16), nn.Conv2d(base_channels*16, 2 * latent_channels, 3, 1, 1))
    def forward(self, x):
        x = self.encoder(x)
        mu, log_var = torch.chunk(x, 2, dim=1)
        return mu, log_var

class DecoderSuperDeep(nn.Module):
    def __init__(self, out_channels=3, base_channels=128, latent_channels=1):
        super().__init__()
        self.decoder = nn.Sequential(nn.Conv2d(latent_channels, base_channels*16, 3, 1, 1), ResBlock(base_channels*16, base_channels*16), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*16, base_channels*8, 3, 1, 1), ResBlock(base_channels*8, base_channels*8), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*8, base_channels*4, 3, 1, 1), ResBlock(base_channels*4, base_channels*4), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*4, base_channels*2, 3, 1, 1), ResBlock(base_channels*2, base_channels*2), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*2, base_channels, 3, 1, 1), ResBlock(base_channels, base_channels), nn.Conv2d(base_channels, out_channels, 3, 1, 1), nn.Tanh())
    def forward(self, z): return self.decoder(z)

class Discriminator(nn.Module):
    def __init__(self, in_channels=3, base_channels=64):
        super().__init__()
        self.model = nn.Sequential(
            nn.Conv2d(in_channels, base_channels, 4, 2, 1), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels, base_channels * 2, 4, 2, 1), nn.BatchNorm2d(base_channels * 2), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels * 2, base_channels * 4, 4, 2, 1), nn.BatchNorm2d(base_channels * 4), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels*4, base_channels*8, 4, 2, 1), nn.BatchNorm2d(base_channels*8), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels * 8, 1, 4, 1, 0) # 최종 출력 크기 조절
        )
    def forward(self, img): return self.model(img)

class InterpolationNet(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM, hidden_dim=1024):
        super().__init__()
        input_dim = latent_dim * 2 + 1
        self.network = nn.Sequential(
            nn.Linear(input_dim, hidden_dim), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim, hidden_dim), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim, hidden_dim // 2), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim // 2, latent_dim)
        )
    def forward(self, z_a, z_b, alpha):
        net_input = torch.cat([z_a, z_b, alpha], dim=1)
        return self.network(net_input)

class CustomImageDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.paths = sorted([os.path.join(root_dir, f) for f in os.listdir(root_dir) if f.lower().endswith(('.jpg','.jpeg','.png'))])
        self.transform = transform
    def __len__(self): return len(self.paths)
    def __getitem__(self, idx):
        img = Image.open(self.paths[idx]).convert('RGB')
        if self.transform: img = self.transform(img)
        return img

def denormalize(tensor):
    return (tensor.clamp(-1, 1) * 0.5) + 0.5

def test_final_network(net, fixed_pairs, encoder, decoder, image_dataset, device):
    # 이전의 test_overfit_network 와 동일한 함수
    print("\n--- Testing the final network ---")
    net.eval()
    for _, (start_idx, end_idx) in enumerate(fixed_pairs):
        num_steps = 9
        alphas = torch.linspace(0, 1, num_steps)
        img_a_orig = image_dataset[start_idx]
        img_b_orig = image_dataset[end_idx]
        with torch.no_grad():
            mu_a, _ = encoder(img_a_orig.unsqueeze(0).to(device))
            mu_b, _ = encoder(img_b_orig.unsqueeze(0).to(device))
        lerp_latents = torch.cat([(1 - alpha) * mu_a + alpha * mu_b for alpha in alphas], dim=0)
        network_latents = []
        with torch.no_grad():
            for alpha in alphas:
                inter_latent_flat = net(mu_a.view(1,-1), mu_b.view(1,-1), torch.tensor([[alpha]], device=device))
                network_latents.append(inter_latent_flat.view(1, 1, 16, 16))
        network_latents = torch.cat(network_latents, dim=0)
        with torch.no_grad():
            lerp_images = decoder(lerp_latents).cpu()
            network_images = decoder(network_latents).cpu()
        fig, axes = plt.subplots(2, num_steps, figsize=(20, 5))
        fig.suptitle(f"Final Comparison (Pair: {start_idx} & {end_idx})", fontsize=16)
        axes[0, 0].set_ylabel("Linear (Lerp)", fontsize=12)
        axes[1, 0].set_ylabel("GAN-Trained Network", fontsize=12)
        for j in range(num_steps):
            axes[0, j].imshow(denormalize(lerp_images[j]).permute(1, 2, 0)); axes[0, j].axis('off')
            axes[1, j].imshow(denormalize(network_images[j]).permute(1, 2, 0)); axes[1, j].axis('off')
            axes[0, j].set_title(f"α={alphas[j]:.2f}")
        plt.show()

# --- 3. 메인 로직 ---
def main():
    # --- 모델 로드 ---
    encoder = nn.DataParallel(EncoderSuperDeep(), device_ids=GPU_IDS).to(DEVICE)
    decoder = nn.DataParallel(DecoderSuperDeep(), device_ids=GPU_IDS).to(DEVICE)
    discriminator = nn.DataParallel(Discriminator(), device_ids=GPU_IDS).to(DEVICE)
    interpolation_net = nn.DataParallel(InterpolationNet(), device_ids=GPU_IDS).to(DEVICE)

    # 사전 학습된 VAE-GAN 가중치 로드
    encoder.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'encoder_superdeep_best.pth'), map_location=DEVICE))
    decoder.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'decoder_superdeep_best.pth'), map_location=DEVICE))
    discriminator.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'discriminator_superdeep_best.pth'), map_location=DEVICE))
    
    # VAE는 고정
    encoder.eval(); decoder.eval()
    for param in encoder.parameters(): param.requires_grad = False
    for param in decoder.parameters(): param.requires_grad = False
    print("Pre-trained VAE & Discriminator models loaded.")

    # --- 데이터셋 생성 ---
    transform = transforms.Compose([transforms.Resize((256, 256)), transforms.ToTensor(), transforms.Normalize(mean=[0.5], std=[0.5])])
    full_image_dataset = CustomImageDataset(DATA_DIR, transform)
    
    # 판별자 학습에 사용할 실제 이미지 데이터 로더
    real_image_subset = Subset(full_image_dataset, range(20000))
    real_image_loader = DataLoader(real_image_subset, batch_size=BATCH_SIZE, shuffle=True)

    # 보간 네트워크 학습용 고정 데이터셋 생성
    print("Generating fixed dataset for interpolation network training...")
    all_net_inputs = []
    for start_idx, end_idx in tqdm(FIXED_PAIRS_INDICES, desc="Generating interp. data"):
        img_a = full_image_dataset[start_idx].unsqueeze(0).to(DEVICE)
        img_b = full_image_dataset[end_idx].unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            mu_a, _ = encoder(img_a)
            mu_b, _ = encoder(img_b)
        alphas = torch.linspace(0.01, 0.99, NUM_POINTS_PER_PAIR).to(DEVICE)
        for alpha in alphas:
            net_input = torch.cat([mu_a.view(1, -1), mu_b.view(1, -1), torch.tensor([[alpha]], device=DEVICE)], dim=1)
            all_net_inputs.append(net_input)
    
    interp_inputs = torch.cat(all_net_inputs, dim=0)
    interp_dataset = TensorDataset(interp_inputs)
    interp_loader = DataLoader(interp_dataset, batch_size=BATCH_SIZE, shuffle=True)
    print(f"Dataset generated. Total interpolation points: {len(interp_dataset)}")

    # --- GAN 방식 훈련 ---
    opt_disc = optim.Adam(discriminator.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999))
    opt_interp = optim.Adam(interpolation_net.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999))
    
    mse_loss_fn = nn.MSELoss()
    adv_loss_fn = nn.BCEWithLogitsLoss()
    
    best_loss = float('inf')
    
    print(f"\nStarting GAN-style overfitting training for {NUM_EPOCHS} epochs...")
    for epoch in range(NUM_EPOCHS):
        interpolation_net.train(); discriminator.train()
        
        # 데이터로더를 동시에 순회하기 위한 이터레이터
        real_iter = iter(real_image_loader)
        
        pbar = tqdm(interp_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")
        for (net_input_batch,) in pbar:
            # --- 1. 판별자 훈련 ---
            opt_disc.zero_grad()
            
            # 실제 이미지 가져오기 (데이터셋이 끝나면 다시 시작)
            try:
                real_imgs = next(real_iter)
            except StopIteration:
                real_iter = iter(real_image_loader)
                real_imgs = next(real_iter)
            real_imgs = real_imgs.to(DEVICE)

            # 보간 네트워크로 가짜 이미지 생성
            z_a = net_input_batch[:, :LATENT_DIM]
            z_b = net_input_batch[:, LATENT_DIM:LATENT_DIM*2]
            alpha = net_input_batch[:, -1].unsqueeze(1)
            
            z_interp_flat = interpolation_net(z_a, z_b, alpha)
            z_interp = z_interp_flat.view(-1, 1, 16, 16)
            fake_imgs = decoder(z_interp)

            # 판별자 손실 계산
            real_pred = discriminator(real_imgs)
            loss_d_real = adv_loss_fn(real_pred, torch.ones_like(real_pred))
            
            fake_pred = discriminator(fake_imgs.detach())
            loss_d_fake = adv_loss_fn(fake_pred, torch.zeros_like(fake_pred))
            
            loss_d = (loss_d_real + loss_d_fake) / 2
            loss_d.backward()
            opt_disc.step()

            # --- 2. 보간 네트워크(생성자) 훈련 ---
            opt_interp.zero_grad()

            # 기하학적 손실 (재투영 오차)
            with torch.no_grad():
                z_hat_interp, _ = encoder(fake_imgs)
            loss_reproj = mse_loss_fn(z_interp, z_hat_interp)
            
            # 적대적 손실
            fake_pred_for_g = discriminator(fake_imgs)
            loss_adv = adv_loss_fn(fake_pred_for_g, torch.ones_like(fake_pred_for_g))
            
            # 최종 손실
            loss_interp = REPROJ_WEIGHT * loss_reproj + ADV_WEIGHT * loss_adv
            loss_interp.backward()
            opt_interp.step()
            
            pbar.set_postfix({
                "D_Loss": f"{loss_d.item():.4f}", 
                "I_Loss": f"{loss_interp.item():.4f}",
                "Reproj": f"{loss_reproj.item():.4f}",
                "Adv": f"{loss_adv.item():.4f}"
            })

        # --- 에폭 종료 후 모델 저장 ---
        if (epoch + 1) % 50 == 0:
            current_loss = loss_interp.item() # 마지막 배치의 손실을 기준으로 저장
            if current_loss < best_loss:
                best_loss = current_loss
                torch.save(interpolation_net.module.state_dict(), INTERP_NET_SAVE_PATH)
                print(f"Epoch {epoch+1}: New best model saved with loss {best_loss:.4f}")

    print("\nTraining finished.")
    
    # --- 4. 최종 결과 테스트 ---
    final_net = nn.DataParallel(InterpolationNet(), device_ids=GPU_IDS).to(DEVICE)
    final_net.module.load_state_dict(torch.load(INTERP_NET_SAVE_PATH, map_location=DEVICE))
    test_final_network(final_net, FIXED_PAIRS_INDICES, encoder, decoder, full_image_dataset, DEVICE)

if __name__ == '__main__':
    main()
```

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset, Subset
import numpy as np
from tqdm.auto import tqdm
from PIL import Image
from torchvision import transforms
import matplotlib.pyplot as plt

# --- 1. 설정 (Settings) ---
DEVICE = torch.device("cuda:2" if torch.cuda.is_available() else "cpu")
GPU_IDS = [2, 1, 3]

# --- 경로 설정 ---
DATA_DIR = '/home/nas/data/YMG/datas/celeba_hq_256/'
AE_CHECKPOINT_DIR = '/home/nas/data/YMG/superdeep_ae/checkpoints/'
SAVE_DIR = '/home/nas/data/YMG/superdeep_ae/보간넷실험04_모드붕괴해결/'
os.makedirs(SAVE_DIR, exist_ok=True)
INTERP_NET_SAVE_PATH = os.path.join(SAVE_DIR, 'interpolation_net_gan_overfit_rebalanced.pth')
LOSS_PLOT_PATH = os.path.join(SAVE_DIR, 'loss_history.png') # 손실 그래프 저장 경로

# --- 하이퍼파라미터 ---
LATENT_DIM = 256
LEARNING_RATE = 1e-5
BATCH_SIZE = 50
NUM_EPOCHS = 1000
REPROJ_WEIGHT = 10.0
ADV_WEIGHT = 1.0

# 테스트에 고정적으로 사용할 이미지 쌍 인덱스
FIXED_PAIRS_INDICES = [
    (9141, 21742), (1857, 17388), (9926, 28731),
    (7732, 1726), (27763, 26175)
]
NUM_POINTS_PER_PAIR = 100

# --- 2. 모델 및 헬퍼 함수 정의 (이전과 동일) ---
class ResBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.block = nn.Sequential(nn.Conv2d(in_channels, out_channels, 3, 1, 1), nn.BatchNorm2d(out_channels), nn.SiLU(), nn.Conv2d(out_channels, out_channels, 3, 1, 1), nn.BatchNorm2d(out_channels))
        self.shortcut = nn.Conv2d(in_channels, out_channels, 1) if in_channels != out_channels else nn.Identity()
        self.silu = nn.SiLU()
    def forward(self, x): return self.silu(self.block(x) + self.shortcut(x))

class EncoderSuperDeep(nn.Module):
    def __init__(self, in_channels=3, base_channels=128, latent_channels=1):
        super().__init__()
        self.encoder = nn.Sequential(nn.Conv2d(in_channels, base_channels, 3, 1, 1), ResBlock(base_channels, base_channels), nn.Conv2d(base_channels, base_channels*2, 3, 2, 1), ResBlock(base_channels*2, base_channels*2), nn.Conv2d(base_channels*2, base_channels*4, 3, 2, 1), ResBlock(base_channels*4, base_channels*4), nn.Conv2d(base_channels*4, base_channels*8, 3, 2, 1), ResBlock(base_channels*8, base_channels*8), nn.Conv2d(base_channels*8, base_channels*16, 3, 2, 1), ResBlock(base_channels*16, base_channels*16), nn.Conv2d(base_channels*16, 2 * latent_channels, 3, 1, 1))
    def forward(self, x):
        x = self.encoder(x)
        mu, log_var = torch.chunk(x, 2, dim=1)
        return mu, log_var

class DecoderSuperDeep(nn.Module):
    def __init__(self, out_channels=3, base_channels=128, latent_channels=1):
        super().__init__()
        self.decoder = nn.Sequential(nn.Conv2d(latent_channels, base_channels*16, 3, 1, 1), ResBlock(base_channels*16, base_channels*16), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*16, base_channels*8, 3, 1, 1), ResBlock(base_channels*8, base_channels*8), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*8, base_channels*4, 3, 1, 1), ResBlock(base_channels*4, base_channels*4), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*4, base_channels*2, 3, 1, 1), ResBlock(base_channels*2, base_channels*2), nn.Upsample(scale_factor=2, mode='nearest'), nn.Conv2d(base_channels*2, base_channels, 3, 1, 1), ResBlock(base_channels, base_channels), nn.Conv2d(base_channels, out_channels, 3, 1, 1), nn.Tanh())
    def forward(self, z): return self.decoder(z)

class Discriminator(nn.Module):
    def __init__(self, in_channels=3, base_channels=64):
        super().__init__()
        self.model = nn.Sequential(
            nn.Conv2d(in_channels, base_channels, 4, 2, 1), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels, base_channels * 2, 4, 2, 1), nn.BatchNorm2d(base_channels * 2), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels * 2, base_channels * 4, 4, 2, 1), nn.BatchNorm2d(base_channels * 4), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels*4, base_channels*8, 4, 2, 1), nn.BatchNorm2d(base_channels*8), nn.LeakyReLU(0.2, inplace=True),
            nn.Conv2d(base_channels * 8, 1, 4, 1, 0)
        )
    def forward(self, img): return self.model(img)

class InterpolationNet(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM, hidden_dim=1024):
        super().__init__()
        input_dim = latent_dim * 2 + 1
        self.network = nn.Sequential(
            nn.Linear(input_dim, hidden_dim), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim, hidden_dim), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim, hidden_dim // 2), nn.LeakyReLU(0.2, inplace=True),
            nn.Linear(hidden_dim // 2, latent_dim)
        )
    def forward(self, z_a, z_b, alpha):
        net_input = torch.cat([z_a, z_b, alpha], dim=1)
        return self.network(net_input)

class CustomImageDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.paths = sorted([os.path.join(root_dir, f) for f in os.listdir(root_dir) if f.lower().endswith(('.jpg','.jpeg','.png'))])
        self.transform = transform
    def __len__(self): return len(self.paths)
    def __getitem__(self, idx):
        img = Image.open(self.paths[idx]).convert('RGB')
        if self.transform: img = self.transform(img)
        return img

def denormalize(tensor):
    return (tensor.clamp(-1, 1) * 0.5) + 0.5

def test_final_network(net, fixed_pairs, encoder, decoder, image_dataset, device):
    print("\n--- Testing the final network ---")
    net.eval()
    for _, (start_idx, end_idx) in enumerate(fixed_pairs):
        num_steps = 9
        alphas = torch.linspace(0, 1, num_steps)
        img_a_orig = image_dataset[start_idx]
        img_b_orig = image_dataset[end_idx]
        with torch.no_grad():
            mu_a, _ = encoder(img_a_orig.unsqueeze(0).to(device))
            mu_b, _ = encoder(img_b_orig.unsqueeze(0).to(device))
        lerp_latents = torch.cat([(1 - alpha) * mu_a + alpha * mu_b for alpha in alphas], dim=0)
        network_latents = []
        with torch.no_grad():
            for alpha in alphas:
                inter_latent_flat = net(mu_a.view(1,-1), mu_b.view(1,-1), torch.tensor([[alpha]], device=device))
                network_latents.append(inter_latent_flat.view(1, 1, 16, 16))
        network_latents = torch.cat(network_latents, dim=0)
        with torch.no_grad():
            lerp_images = decoder(lerp_latents).cpu()
            network_images = decoder(network_latents).cpu()
        fig, axes = plt.subplots(2, num_steps, figsize=(20, 5))
        fig.suptitle(f"Final Comparison (Pair: {start_idx} & {end_idx})", fontsize=16)
        axes[0, 0].set_ylabel("Linear (Lerp)", fontsize=12)
        axes[1, 0].set_ylabel("GAN-Trained Network", fontsize=12)
        for j in range(num_steps):
            axes[0, j].imshow(denormalize(lerp_images[j]).permute(1, 2, 0)); axes[0, j].axis('off')
            axes[1, j].imshow(denormalize(network_images[j]).permute(1, 2, 0)); axes[1, j].axis('off')
            axes[0, j].set_title(f"α={alphas[j]:.2f}")
        plt.show()

# <<< [추가됨] 손실 그래프 시각화 함수 >>>
def plot_loss_history(history, save_path):
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle('Training Loss History', fontsize=16) # 타이틀: 학습 손실 기록

    # 판별자 손실
    axes[0, 0].plot(history['d_loss'], label='Discriminator Loss', color='r')
    axes[0, 0].set_title('Discriminator Loss') # 그래프 제목: 판별자 손실
    axes[0, 0].set_xlabel('Epochs') # x축 이름: 에폭
    axes[0, 0].set_ylabel('Loss') # y축 이름: 손실
    axes[0, 0].legend()

    # 보간 네트워크 총 손실
    axes[0, 1].plot(history['i_loss'], label='Interpolator Loss', color='b')
    axes[0, 1].set_title('Interpolator Total Loss') # 그래프 제목: 보간 네트워크 총 손실
    axes[0, 1].set_xlabel('Epochs')
    axes[0, 1].set_ylabel('Loss')
    axes[0, 1].legend()

    # 재투영 오차
    axes[1, 0].plot(history['reproj_loss'], label='Reprojection Loss', color='g')
    axes[1, 0].set_title('Reprojection Loss') # 그래프 제목: 재투영 오차
    axes[1, 0].set_xlabel('Epochs')
    axes[1, 0].set_ylabel('Loss')
    axes[1, 0].legend()

    # 적대적 손실
    axes[1, 1].plot(history['adv_loss'], label='Adversarial Loss', color='purple')
    axes[1, 1].set_title('Adversarial Loss') # 그래프 제목: 적대적 손실
    axes[1, 1].set_xlabel('Epochs')
    axes[1, 1].set_ylabel('Loss')
    axes[1, 1].legend()

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(save_path)
    print(f"\nLoss history plot saved to {save_path}")
    plt.show()
# <<< 추가 끝 >>>


# --- 3. 메인 로직 ---
def main():
    # --- 모델 로드 ---
    encoder = nn.DataParallel(EncoderSuperDeep(), device_ids=GPU_IDS).to(DEVICE)
    decoder = nn.DataParallel(DecoderSuperDeep(), device_ids=GPU_IDS).to(DEVICE)
    discriminator = nn.DataParallel(Discriminator(), device_ids=GPU_IDS).to(DEVICE)
    interpolation_net = nn.DataParallel(InterpolationNet(), device_ids=GPU_IDS).to(DEVICE)

    encoder.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'encoder_superdeep_best.pth'), map_location=DEVICE))
    decoder.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'decoder_superdeep_best.pth'), map_location=DEVICE))
    discriminator.module.load_state_dict(torch.load(os.path.join(AE_CHECKPOINT_DIR, 'discriminator_superdeep_best.pth'), map_location=DEVICE))
    
    encoder.eval(); decoder.eval()
    for param in encoder.parameters(): param.requires_grad = False
    for param in decoder.parameters(): param.requires_grad = False
    print("Pre-trained VAE & Discriminator models loaded.")

    # --- 데이터셋 생성 ---
    transform = transforms.Compose([transforms.Resize((256, 256)), transforms.ToTensor(), transforms.Normalize(mean=[0.5], std=[0.5])])
    full_image_dataset = CustomImageDataset(DATA_DIR, transform)
    
    real_image_subset = Subset(full_image_dataset, range(20000))
    real_image_loader = DataLoader(real_image_subset, batch_size=BATCH_SIZE, shuffle=True)

    print("Generating fixed dataset for interpolation network training...")
    all_net_inputs = []
    for start_idx, end_idx in tqdm(FIXED_PAIRS_INDICES, desc="Generating interp. data"):
        img_a = full_image_dataset[start_idx].unsqueeze(0).to(DEVICE)
        img_b = full_image_dataset[end_idx].unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            mu_a, _ = encoder(img_a)
            mu_b, _ = encoder(img_b)
        alphas = torch.linspace(0.01, 0.99, NUM_POINTS_PER_PAIR).to(DEVICE)
        for alpha in alphas:
            net_input = torch.cat([mu_a.view(1, -1), mu_b.view(1, -1), torch.tensor([[alpha]], device=DEVICE)], dim=1)
            all_net_inputs.append(net_input)
    
    interp_inputs = torch.cat(all_net_inputs, dim=0)
    interp_dataset = TensorDataset(interp_inputs)
    interp_loader = DataLoader(interp_dataset, batch_size=BATCH_SIZE, shuffle=True)
    print(f"Dataset generated. Total interpolation points: {len(interp_dataset)}")

    # --- GAN 방식 훈련 ---
    opt_disc = optim.Adam(discriminator.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999))
    opt_interp = optim.Adam(interpolation_net.parameters(), lr=LEARNING_RATE, betas=(0.5, 0.999))
    
    mse_loss_fn = nn.MSELoss()
    adv_loss_fn = nn.BCEWithLogitsLoss()
    
    best_loss = float('inf')
    
    # <<< [추가됨] 손실 기록을 위한 딕셔너리 >>>
    loss_history = {'d_loss': [], 'i_loss': [], 'reproj_loss': [], 'adv_loss': []}
    
    print(f"\nStarting GAN-style overfitting training for {NUM_EPOCHS} epochs...")
    for epoch in range(NUM_EPOCHS):
        interpolation_net.train(); discriminator.train()
        
        # <<< [추가됨] 에폭별 손실 누적 변수 >>>
        total_d, total_i, total_reproj, total_adv = 0, 0, 0, 0
        
        real_iter = iter(real_image_loader)
        pbar = tqdm(interp_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")
        for (net_input_batch,) in pbar:
            # --- 1. 판별자 훈련 ---
            opt_disc.zero_grad()
            try:
                real_imgs = next(real_iter)
            except StopIteration:
                real_iter = iter(real_image_loader)
                real_imgs = next(real_iter)
            real_imgs = real_imgs.to(DEVICE)

            z_a = net_input_batch[:, :LATENT_DIM]
            z_b = net_input_batch[:, LATENT_DIM:LATENT_DIM*2]
            alpha = net_input_batch[:, -1].unsqueeze(1)
            
            z_interp_flat = interpolation_net(z_a, z_b, alpha)
            z_interp = z_interp_flat.view(-1, 1, 16, 16)
            fake_imgs = decoder(z_interp)

            real_pred = discriminator(real_imgs)
            loss_d_real = adv_loss_fn(real_pred, torch.ones_like(real_pred))
            fake_pred = discriminator(fake_imgs.detach())
            loss_d_fake = adv_loss_fn(fake_pred, torch.zeros_like(fake_pred))
            loss_d = (loss_d_real + loss_d_fake) / 2
            loss_d.backward()
            opt_disc.step()

            # --- 2. 보간 네트워크(생성자) 훈련 ---
            opt_interp.zero_grad()
            with torch.no_grad():
                z_hat_interp, _ = encoder(fake_imgs)
            loss_reproj = mse_loss_fn(z_interp, z_hat_interp)
            fake_pred_for_g = discriminator(fake_imgs)
            loss_adv = adv_loss_fn(fake_pred_for_g, torch.ones_like(fake_pred_for_g))
            loss_interp = REPROJ_WEIGHT * loss_reproj + ADV_WEIGHT * loss_adv
            loss_interp.backward()
            opt_interp.step()
            
            # <<< [추가됨] 배치 손실 누적 >>>
            total_d += loss_d.item()
            total_i += loss_interp.item()
            total_reproj += loss_reproj.item()
            total_adv += loss_adv.item()
            
            pbar.set_postfix({
                "D_Loss": f"{loss_d.item():.4f}", "I_Loss": f"{loss_interp.item():.4f}",
                "Reproj": f"{loss_reproj.item():.4f}", "Adv": f"{loss_adv.item():.4f}"
            })

        # <<< [추가됨] 에폭 평균 손실 계산 및 기록 >>>
        num_batches = len(interp_loader)
        loss_history['d_loss'].append(total_d / num_batches)
        loss_history['i_loss'].append(total_i / num_batches)
        loss_history['reproj_loss'].append(total_reproj / num_batches)
        loss_history['adv_loss'].append(total_adv / num_batches)
        
        # 에폭 종료 후 모델 저장
        current_loss = total_i / num_batches
        if current_loss < best_loss:
            best_loss = current_loss
            torch.save(interpolation_net.module.state_dict(), INTERP_NET_SAVE_PATH)
            print(f"Epoch {epoch+1}: New best model saved with loss {best_loss:.4f}")

    print("\nTraining finished.")
    
    # --- 4. 최종 결과 테스트 및 시각화 ---
    final_net = nn.DataParallel(InterpolationNet(), device_ids=GPU_IDS).to(DEVICE)
    final_net.module.load_state_dict(torch.load(INTERP_NET_SAVE_PATH, map_location=DEVICE))
    test_final_network(final_net, FIXED_PAIRS_INDICES, encoder, decoder, full_image_dataset, DEVICE)
    
    # <<< [추가됨] 손실 그래프 그리기 >>>
    plot_loss_history(loss_history, LOSS_PLOT_PATH)

if __name__ == '__main__':
    main()
```

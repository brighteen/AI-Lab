> 
> 응용 분야에서 선형대수학의 중요성은 컴퓨팅 파워의 증가에 정비례하여 상승해왔으며, 새로운 세대의 하드웨어와 소프트웨어가 나올 때마다 훨씬 더 큰 능력을 요구하게 되었습니다. 따라서 컴퓨터 과학은 병렬 처리와 대규모 계산의 폭발적인 성장을 통해 선형대수학과 복잡하게 연결되어 있습니다.
> 
> 과학자들과 공학자들은 이제 수십 년 전에는 상상조차 할 수 없었던 훨씬 더 복잡한 문제들을 다루고 있습니다. 오늘날 선형대수학은 많은 과학 및 비즈니스 분야의 학생들에게 다른 어떤 학부 수학 과목보다 더 큰 잠재적 가치를 가지고 있습니다! 이 교재의 내용은 많은 흥미로운 분야에서의 추가적인 작업을 위한 기초를 제공합니다. 몇 가지 가능성은 다음과 같으며, 다른 것들은 나중에 설명될 것입니다.
> 
> * **석유 탐사.** 배가 해상 석유 매장지를 탐사할 때, 컴퓨터는 매일 수천 개의 개별적인 선형방정식 시스템을 풉니다. 방정식에 대한 지진 데이터는 에어건의 폭발로 생성된 수중 충격파로부터 얻어집니다. 파동은 지하 암석에서 반사되어 배 뒤에 있는 수 마일 길이의 케이블에 부착된 지오폰(geophone)으로 측정됩니다.
> * **선형 계획법.** 오늘날 많은 중요한 경영 결정은 수백 개의 변수를 사용하는 선형 계획법 모델을 기반으로 이루어집니다. 예를 들어, 항공 산업은 비행 승무원 일정을 짜고, 항공기 위치를 모니터링하며, 유지보수 및 터미널 운영과 같은 다양한 지원 서비스의 일정을 계획하는 선형 프로그램을 사용합니다.
> * **전기 회로망.** 공학자들은 수백만 개의 트랜지스터를 포함하는 전기 회로와 마이크로칩을 설계하기 위해 시뮬레이션 소프트웨어를 사용합니다. 이러한 소프트웨어는 선형대수학 기법과 선형방정식 시스템에 의존합니다.
> * **인공 지능.** 선형대수학은 데이터 정제에서부터 안면 인식에 이르기까지 모든 것에서 핵심적인 역할을 합니다.
> * **신호와 신호 처리.** 디지털 사진에서부터 주식의 일일 가격에 이르기까지, 중요한 정보는 신호로 기록되고 선형 변환을 사용하여 처리됩니다.
> * **기계 학습.** 기계(특히 컴퓨터)는 온라인 쇼핑 선호도에서부터 음성 인지에 이르기까지 모든 것을 학습하기 위해 선형대수학을 사용합니다.

선형방정식 시스템은 선형대수학의 중심에 있으며, 이 장에서는 이를 사용하여 선형대수학의 몇 가지 핵심 개념을 간단하고 구체적인 환경에서 소개합니다. 1.1절과 1.2절에서는 선형방정식 시스템을 풀기 위한 체계적인 방법을 제시합니다. 이 알고리즘은 교재 전체의 계산에 사용될 것입니다. 1.3절과 1.4절에서는 선형방정식 시스템이 벡터방정식 및 행렬방정식과 어떻게 동등한지를 보여줍니다. 이 동등성은 벡터의 선형 결합과 관련된 문제를 선형방정식 시스템에 대한 질문으로 환원시킬 것입니다. 이 장의 후반부에서 연구될 생성(spanning), 선형독립(linear independence), 그리고 선형 변환(linear transformation)의 기본 개념들은 우리가 선형대수학의 아름다움과 힘을 탐구함에 따라 교재 전체에서 필수적인 역할을 할 것입니다.

## 1.1 선형방정식 시스템 (Systems of Linear Equations)

변수 $x_{1}, \dots, x_{n}$에 대한 **선형방정식(linear equation)**은 다음과 같은 형태로 쓰일 수 있는 방정식이다.

$$a_{1}x_{1} + a_{2}x_{2} + \dots + a_{n}x_{n} = b \quad (1)$$

여기서 $b$와 계수 $a_{1}, \dots, a_{n}$은 보통 미리 알려진 실수 또는 복소수이다. 아래 첨자 $n$은 어떤 양의 정수도 될 수 있다. 교재의 예제와 연습문제에서 $n$은 보통 2와 5 사이이다. 실제 문제에서 $n$은 50 또는 5000, 혹은 그보다 더 클 수 있다.

다음 방정식들은

$$4x_{1} - 5x_{2} + 2 = x_{1} \quad \text{그리고} \quad x_{2} = 2(\sqrt{6} - x_{1}) + x_{3}$$

방정식 (1)의 형태로 대수적으로 재배열될 수 있기 때문에 모두 선형이다:

$$3x_{1} - 5x_{2} = -2 \quad \text{그리고} \quad 2x_{1} + x_{2} - x_{3} = 2\sqrt{6}$$

다음 방정식들은

$$4x_{1} - 5x_{2} = x_{1}x_{2} \quad \text{그리고} \quad x_{2} = 2\sqrt{x_{1}} - 6$$

첫 번째 방정식에 $x_{1}x_{2}$가 있고 두 번째 방정식에 $\sqrt{x_{1}}$가 있기 때문에 선형이 아니다.

**선형방정식 시스템(system of linear equations)** (또는 **선형 시스템(linear system)**)은 $x_{1}, \dots, x_{n}$과 같이 동일한 변수들을 포함하는 하나 이상의 선형방정식의 모음이다. 한 예는 다음과 같다.

$$2x_{1} - x_{2} + 1.5x_{3} = 8$$
$$x_{1} - 4x_{3} = -7 \quad (2)$$

시스템의 **해(solution)**는 각 방정식을 참인 명제로 만드는 수들의 목록 $(s_{1}, s_{2}, \dots, s_{n})$이다. 이때 값 $s_{1}, \dots, s_{n}$은 각각 $x_{1}, \dots, x_{n}$에 대입된다. 예를 들어, $(5, 6.5, 3)$은 시스템 (2)의 해이다. 왜냐하면 이 값들을 (2)에서 각각 $x_{1}, x_{2}, x_{3}$에 대입하면, 방정식은 $8=8$과 $-7=-7$로 단순화되기 때문이다.

모든 가능한 해의 집합을 선형 시스템의 **해집합(solution set)**이라고 한다. 두 선형 시스템이 동일한 해집합을 가질 때 **동치(equivalent)**라고 한다. 즉, 첫 번째 시스템의 각 해는 두 번째 시스템의 해이고, 두 번째 시스템의 각 해는 첫 번째 시스템의 해이다.

두 변수를 가진 두 개의 선형방정식 시스템의 해집합을 찾는 것은 두 직선의 교점을 찾는 것과 같기 때문에 쉽다. 일반적인 문제는 다음과 같다.

$$x_{1} - 2x_{2} = -1$$
$$-x_{1} + 3x_{2} = 3$$

이 방정식들의 그래프는 직선이며, 이를 각각 $l_{1}$과 $l_{2}$로 나타낸다. 수의 쌍 $(x_{1}, x_{2})$가 시스템의 두 방정식을 모두 만족시키는 것은 점 $(x_{1}, x_{2})$가 $l_{1}$과 $l_{2}$ 위에 모두 있을 때, 그리고 오직 그 때만이다. 위 시스템에서 해는 단일 점 $(3, 2)$이며, 이는 쉽게 확인할 수 있다. (그림 1 참조)

```python
import numpy as np
import matplotlib.pyplot as plt

# x1 값의 범위를 정의한다.
x1 = np.linspace(-1, 5, 100)

# 두 개의 선형방정식을 정의한다.
# l1: x1 - 2x2 = -1  => x2 = (x1 + 1) / 2
x2_l1 = (x1 + 1) / 2
# l2: -x1 + 3x2 = 3 => x2 = (x1 + 3) / 3
x2_l2 = (x1 + 3) / 3

# 그래프를 생성한다.
plt.figure(figsize=(6, 5))
plt.plot(x1, x2_l1, label='$x_1 - 2x_2 = -1$ ($l_1$)')
plt.plot(x1, x2_l2, label='$-x_1 + 3x_2 = 3$ ($l_2$)')

# 교점을 계산한다.
A = np.array([[1, -2], [-1, 3]])
b = np.array([-1, 3])
intersection = np.linalg.solve(A, b)

# 교점을 그래프에 표시한다.
plt.plot(intersection[0], intersection[1], 'ro') # ro는 빨간색 원 마커를 의미한다.
plt.text(intersection[0] + 0.1, intersection[1], f'({intersection[0]:.0f}, {intersection[1]:.0f})', fontsize=12)

# 축과 레이블을 설정한다.
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('FIGURE 1: Exactly one solution')
plt.grid(True)
plt.legend()
plt.axis('equal')
plt.show()
```

물론, 두 직선이 반드시 한 점에서 만나는 것은 아니다. 평행할 수도 있고, 일치하여 직선 위의 모든 점에서 "만날" 수도 있다. 그림 2는 다음 시스템에 해당하는 그래프를 보여준다.

**(a)**
$$x_{1} - 2x_{2} = -1$$
$$-x_{1} + 2x_{2} = 3$$

**(b)**
$$x_{1} - 2x_{2} = -1$$
$$-x_{1} + 2x_{2} = 1$$

그림 1과 2는 1.2절에서 증명될 선형 시스템에 대한 다음과 같은 일반적인 사실을 보여준다.

```python
import numpy as np
import matplotlib.pyplot as plt

# x1 값의 범위를 정의한다.
x1 = np.linspace(-2, 4, 100)

# 서브플롯을 생성한다.
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# --- (a) 해가 없는 경우 --- 
# l1: x1 - 2x2 = -1 => x2 = (x1 + 1) / 2
x2_a1 = (x1 + 1) / 2
# l2: -x1 + 2x2 = 3 => x2 = (x1 + 3) / 2
x2_a2 = (x1 + 3) / 2
ax1.plot(x1, x2_a1, label='$l_1: x_1 - 2x_2 = -1$')
ax1.plot(x1, x2_a2, label='$l_2: -x_1 + 2x_2 = 3$')
ax1.set_title('(a) No solution')
ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.grid(True)
ax1.legend()
ax1.axis('equal')

# --- (b) 무한히 많은 해를 갖는 경우 --- 
# l1: x1 - 2x2 = -1 => x2 = (x1 + 1) / 2
x2_b1 = (x1 + 1) / 2
# l2: -x1 + 2x2 = 1 => x2 = (x1 + 1) / 2
x2_b2 = (x1 + 1) / 2
ax2.plot(x1, x2_b1, label='$l_1, l_2: x_1 - 2x_2 = -1$')
ax2.set_title('(b) Infinitely many solutions')
ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.grid(True)
ax2.legend()
ax2.axis('equal')

plt.suptitle('FIGURE 2')
plt.show()
```

> 선형방정식 시스템은 다음 중 하나를 만족한다.
> 1. 해가 없다(no solution).
> 2. 정확히 하나의 해를 갖는다(exactly one solution).
> 3. 무한히 많은 해를 갖는다(infinitely many solutions).

선형방정식 시스템이 하나의 해 또는 무한히 많은 해를 가질 때 **해가 존재한다(consistent)**고 말하며, 해가 없을 때 **해가 존재하지 않는다(inconsistent)**고 말한다.

### 행렬 표기법 (Matrix Notation)

선형 시스템의 필수 정보는 **행렬(matrix)**이라고 불리는 직사각형 배열에 간결하게 기록될 수 있다. 다음 시스템이 주어졌을 때,

$$x_{1} - 2x_{2} + x_{3} = 0$$
$$2x_{2} - 8x_{3} = 8$$
$$5x_{1} - 5x_{3} = 10 \quad (3)$$

각 변수의 계수가 열에 맞추어 정렬되어 있을 때, 행렬

$$
\begin{bmatrix}
1 & -2 & 1 \\
0 & 2 & -8 \\
5 & 0 & -5
\end{bmatrix}
$$

을 시스템 (3)의 **계수 행렬(coefficient matrix)**이라고 하며, 행렬

$$
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 2 & -8 & 8 \\
5 & 0 & -5 & 10
\end{bmatrix} \quad (4)
$$

을 시스템의 **첨가 행렬(augmented matrix)**이라고 한다. (여기서 두 번째 행은 두 번째 방정식이 $0 \cdot x_{1} + 2x_{2} - 8x_{3} = 8$로 쓰일 수 있기 때문에 0을 포함한다.) 시스템의 첨가 행렬은 계수 행렬에 방정식의 각 우변 상수들을 포함하는 열을 추가하여 구성된다.

행렬의 **크기(size)**는 행과 열이 몇 개인지를 나타낸다. 위 첨가 행렬 (4)는 3개의 행과 4개의 열을 가지며, $3 \times 4$ ("3 by 4"로 읽는다) 행렬이라고 한다. $m$과 $n$이 양의 정수일 때, $m \times n$ 행렬은 $m$개의 행과 $n$개의 열을 가진 수들의 직사각형 배열이다. (행의 수가 항상 먼저 온다.) 행렬 표기법은 이어지는 예제들의 계산을 단순화할 것이다.

### 선형 시스템 풀이 (Solving a Linear System)

이 절과 다음 절에서는 선형 시스템을 풀기 위한 **알고리즘(algorithm)**, 즉 체계적인 절차를 설명한다. 기본 전략은 하나의 시스템을 풀기 더 쉬운 **동치 시스템(equivalent system)**(즉, 동일한 해집합을 갖는 시스템)으로 바꾸는 것이다.

대략적으로 말하면, 시스템의 첫 번째 방정식에 있는 $x_{1}$ 항을 사용하여 다른 방정식들의 $x_{1}$ 항을 제거한다. 그런 다음 두 번째 방정식에 있는 $x_{2}$ 항을 사용하여 다른 방정식들의 $x_{2}$ 항을 제거하는 식으로, 마침내 매우 간단한 동치 방정식 시스템을 얻을 때까지 계속한다.

선형 시스템을 단순화하기 위해 세 가지 기본 연산이 사용된다: 한 방정식을 그 자신과 다른 방정식의 배수를 더한 것으로 교체하고, 두 방정식을 서로 바꾸며, 한 방정식의 모든 항에 0이 아닌 상수를 곱한다. 첫 번째 예제 이후에, 이 세 연산이 왜 시스템의 해집합을 바꾸지 않는지 알게 될 것이다.

### 📝 예제 1 (Example 1)

시스템 (3)을 풀어라.

**해법**
소거 절차는 여기에 행렬 표기법과 함께, 그리고 행렬 표기법 없이 나란히 비교하여 보여준다:

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
2x_{2} - 8x_{3} &= 8 \\ 
5x_{1} - 5x_{3} &= 10
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 2 & -8 & 8 \\
5 & 0 & -5 & 10
\end{bmatrix}
$$

$x_{1}$을 첫 번째 방정식에 유지하고 다른 방정식들에서 제거한다. 이를 위해 방정식 1에 -5를 곱한 것을 방정식 3에 더한다. 연습을 좀 하면, 이런 종류의 계산은 보통 암산으로 수행된다:

$$
\begin{alignedat}{7}
-5 \times \text{[방정식 1]} \quad && -5x_{1} &+& 10x_{2} &-& 5x_{3} &= 0 \\ 
+ \text{[방정식 3]} \quad && 5x_{1} & & &-& 5x_{3} &= 10 \\ 
\hline
\text{[새 방정식 3]} \quad && & & 10x_{2} &-& 10x_{3} &= 10
\end{alignedat}
$$

이 계산의 결과는 원래의 세 번째 방정식 대신에 쓰여진다:

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
2x_{2} - 8x_{3} &= 8 \\ 
10x_{2} - 10x_{3} &= 10
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 2 & -8 & 8 \\
0 & 10 & -10 & 10
\end{bmatrix}
$$

이제, 방정식 2를 $\frac{1}{2}$로 곱하여 $x_{2}$의 계수를 1로 만든다. (이 계산은 다음 단계의 산수를 단순화할 것이다.)

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
x_{2} - 4x_{3} &= 4 \\ 
10x_{2} - 10x_{3} &= 10
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 1 & -4 & 4 \\
0 & 10 & -10 & 10
\end{bmatrix}
$$

방정식 2의 $x_{2}$를 사용하여 방정식 3의 $10x_{2}$를 제거한다. "암산" 계산은 다음과 같다.

$$
\begin{alignedat}{7}
-10 \times \text{[방정식 2]} \quad && -10x_{2} &+& 40x_{3} && &= -40 \\ 
+ \text{[방정식 3]} \quad && 10x_{2} &-& 10x_{3} && &= 10 \\ 
\hline
\text{[새 방정식 3]} \quad && & & 30x_{3} && &= -30
\end{alignedat}
$$

이 계산의 결과는 이전의 세 번째 방정식 (행) 대신에 쓰여진다:

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
x_{2} - 4x_{3} &= 4 \\ 
30x_{3} &= -30
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 1 & -4 & 4 \\
0 & 0 & 30 & -30
\end{bmatrix}
$$

이제, 방정식 3을 $\frac{1}{30}$으로 곱하여 $x_{3}$의 계수를 1로 만든다. (이 계산은 다음 단계의 산수를 단순화할 것이다.)

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
x_{2} - 4x_{3} &= 4 \\ 
x_{3} &= -1
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 1 & -4 & 4 \\
0 & 0 & 1 & -1
\end{bmatrix}
$$

새로운 시스템은 삼각 형태(triangular form)를 가진다. (직관적인 용어 '삼각'은 다음 절에서 정확한 용어로 대체될 것이다.)

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} + x_{3} &= 0 \\ 
x_{2} - 4x_{3} &= 4 \\ 
x_{3} &= -1
\end{alignedat}
$$

궁극적으로는 방정식 1에서 $-2x_{2}$ 항을 제거하고 싶지만, 방정식 3의 $x_{3}$을 먼저 사용하여 방정식 2와 1의 $-4x_{3}$과 $+x_{3}$ 항을 제거하는 것이 더 효율적이다. 두 "암산" 계산은 다음과 같다.

$$
\begin{matrix}
4 \times \text{[eq 3]} & 4x_3 = -4 & & -1 \times \text{[eq 3]} & -x_3=1 \\
+ \text{[eq 2]} & x_2 - 4x_3 = 4 & & + \text{[eq 1]} & x_1 - 2x_2 + x_3=0 \\
\hline
\text{[new eq 2]} & x_2 = 0 & & \text{[new eq 1]} & x_1 - 2x_2 = 1
\end{matrix}
$$

이 두 연산의 결과를 결합하는 것이 편리하다:

$$
\begin{alignedat}{2}
x_{1} - 2x_{2} &= 1 \\ 
x_{2} &= 0 \\ 
x_{3} &= -1
\end{alignedat}
\qquad
\begin{bmatrix}
1 & -2 & 0 & 1 \\
0 & 1 & 0 & 0 \\
0 & 0 & 1 & -1
\end{bmatrix}
$$

이제 방정식 3 위의 열을 정리했으므로, 방정식 2의 $x_{2}$로 돌아가서 그 위의 $-2x_{2}$를 제거한다. 이전의 $x_{3}$ 작업 덕분에 이제 $x_{3}$ 항과 관련된 산수는 없다. 방정식 2에 2를 곱한 것을 방정식 1에 더하면 다음 시스템을 얻는다:

$$
\begin{alignedat}{2}
x_{1} &= 1 \\ 
x_{2} &= 0 \\ 
x_{3} &= -1
\end{alignedat}
\qquad
\begin{bmatrix}
1 & 0 & 0 & 1 \\
0 & 1 & 0 & 0 \\
0 & 0 & 1 & -1
\end{bmatrix}
$$

작업은 본질적으로 끝났다. 이는 원래 시스템의 유일한 해가 $(1, 0, -1)$임을 보여준다. 그러나 계산이 많기 때문에 작업을 확인하는 것이 좋은 습관이다. $(1, 0, -1)$이 해임을 확인하기 위해, 이 값들을 원래 시스템의 좌변에 대입하고 계산한다:

$$1(1) - 2(0) + 1(-1) = 1 - 0 - 1 = 0$$
$$2(0) - 8(-1) = 0 + 8 = 8$$
$$5(1) - 5(-1) = 5 + 5 = 10$$

결과는 원래 시스템의 우변과 일치하므로 $(1, 0, -1)$은 시스템의 해이다.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 3차원 그래프를 위한 설정
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# x1, x2 값의 범위를 생성한다.
x1 = np.linspace(-2, 4, 10)
x2 = np.linspace(-2, 4, 10)
x1, x2 = np.meshgrid(x1, x2)

# 세 평면의 방정식을 정의한다.
# 1: x1 - 2x2 + x3 = 0  => x3 = -x1 + 2x2
x3_1 = -x1 + 2*x2
# 2: 2x2 - 8x3 = 8 => x3 = (2x2 - 8) / 8 = 0.25*x2 - 1
x3_2 = 0.25*x2 - 1
# 3: 5x1 - 5x3 = 10 => x3 = (5x1 - 10) / 5 = x1 - 2
x3_3 = x1 - 2

# 각 평면을 그린다.
ax.plot_surface(x1, x2, x3_1, alpha=0.5, rstride=100, cstride=100, color='r')
ax.plot_surface(x1, x2, x3_2, alpha=0.5, rstride=100, cstride=100, color='g')
ax.plot_surface(x1, x2, x3_3, alpha=0.5, rstride=100, cstride=100, color='b')

# 교점 (1, 0, -1)을 표시한다.
ax.scatter([1], [0], [-1], color='black', s=100, label='Intersection (1, 0, -1)')

# 축 레이블을 설정한다.
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$x_3$')
ax.set_title('Intersection of Three Planes')
ax.legend()

plt.show()
```

예제 1은 선형 시스템의 방정식에 대한 연산이 첨가 행렬의 해당 행에 대한 연산과 어떻게 일치하는지를 보여준다. 앞서 나열된 세 가지 기본 연산은 첨가 행렬에 대한 다음 연산에 해당한다.

> ### 기본 행 연산 (ELEMENTARY ROW OPERATIONS)
> 1. (**교체, Replacement**) 한 행을 그 자신과 다른 행의 배수를 더한 것으로 교체한다.
> 2. (**교환, Interchange**) 두 행을 서로 바꾼다.
> 3. (**스케일링, Scaling**) 한 행의 모든 성분에 0이 아닌 상수를 곱한다.

행 연산은 선형 시스템의 첨가 행렬로 나타나는 행렬뿐만 아니라 어떤 행렬에도 적용될 수 있다. 두 행렬은 하나의 행렬을 다른 행렬로 변환하는 일련의 기본 행 연산이 있을 경우 **행 동치(row equivalent)**라고 한다.

행 연산은 **가역적(reversible)**이라는 점에 주목하는 것이 중요하다. 만약 두 행이 교환되었다면, 또 다른 교환으로 원래 위치로 되돌릴 수 있다. 만약 한 행이 0이 아닌 상수 $c$로 스케일링되었다면, 새 행을 $1/c$로 곱하면 원래 행이 생성된다. 마지막으로, 두 행(가령 1행과 2행)을 포함하는 교체 연산을 생각해보자. 1행에 $c$를 곱한 것을 2행에 더하여 새로운 2행을 만들었다고 가정하자. 이 연산을 "되돌리려면", 1행에 $-c$를 곱한 것을 (새로운) 2행에 더하면 원래의 2행을 얻게 된다.

현재 우리는 선형방정식 시스템의 첨가 행렬에 대한 행 연산에 관심이 있다. 어떤 시스템이 행 연산을 통해 새로운 시스템으로 변경되었다고 가정해보자. 각 유형의 행 연산을 고려하면, 원래 시스템의 어떤 해도 새로운 시스템의 해로 남아 있음을 알 수 있다. 역으로, 원래 시스템은 새로운 시스템에 대한 행 연산을 통해 생성될 수 있으므로, 새로운 시스템의 각 해도 원래 시스템의 해이다. 이 논의는 다음의 명제를 정당화한다.

> 두 선형 시스템의 첨가 행렬이 행 동치라면, 두 시스템은 동일한 해집합을 가진다.

예제 1은 길지만, 연습을 좀 하면 계산이 빠르게 진행될 것이다. 이 교재와 연습문제의 행 연산은 보통 수행하기 매우 쉬워서, 근본적인 개념에 집중할 수 있게 해준다. 그럼에도 불구하고, 행 연산은 교재 전체에서 사용될 것이므로 정확하게 수행하는 법을 배워야 한다.

이 절의 나머지 부분에서는 선형 시스템을 완전히 풀지 않고도 해집합의 크기를 결정하기 위해 행 연산을 사용하는 방법을 보여준다.

### 존재성과 유일성 질문 (Existence and Uniqueness Questions)

1.2절에서는 선형 시스템의 해집합이 왜 해가 없거나, 하나의 해를 갖거나, 또는 무한히 많은 해를 갖는지를 보여줄 것이다. 다음 두 질문에 대한 답은 선형 시스템의 해집합의 성격을 결정할 것이다.

특정 시스템에 대해 어떤 가능성이 참인지 결정하기 위해, 우리는 두 가지 질문을 한다.

> ### 선형 시스템에 대한 두 가지 근본적인 질문
> 1. 시스템이 **해가 존재하는가(consistent)**; 즉, 적어도 하나의 해가 존재하는가?
> 2. 만약 해가 존재한다면, 그것이 유일한 해인가; 즉, 해가 **유일한가(unique)**?

이 두 질문은 교재 전반에 걸쳐 다양한 형태로 나타날 것이다. 이 절과 다음 절에서는 첨가 행렬에 대한 행 연산을 통해 이러한 질문에 답하는 방법을 보여줄 것이다.

### 📝 예제 2 (Example 2)

다음 시스템이 해를 가지는지 판별하라:

$$
\begin{alignedat}{4}
x_1 & - & 2x_2 & + & x_3 & = & 0 \\
& & 2x_2 & - & 8x_3 & = & 8 \\
5x_1 & & & - & 5x_3 & = & 10
\end{alignedat}
$$

**해법**

이것은 예제 1의 시스템이다. 삼각 형태를 얻기 위해 필요한 행 연산을 수행했다고 가정하자.

$$
\begin{alignedat}{4}
x_1 & - & 2x_2 & + & x_3 & = & 0 \\
& & x_2 & - & 4x_3 & = & 4 \\
& & & & x_3 & = & -1
\end{alignedat}
\quad
\begin{bmatrix}
1 & -2 & 1 & 0 \\
0 & 1 & -4 & 4 \\
0 & 0 & 1 & -1
\end{bmatrix}
$$

이 시점에서 우리는 $x_3$를 안다. $x_3$의 값을 방정식 2에 대입하면 $x_2$를 계산할 수 있고, 따라서 방정식 1로부터 $x_1$을 결정할 수 있다. 그러므로 해가 존재한다; 시스템은 해를 가진다(consistent). (사실, $x_3$가 오직 하나의 가능한 값을 가지므로 $x_2$는 방정식 2에 의해 유일하게 결정되고, 따라서 $x_1$은 방정식 1에 의해 유일하게 결정된다. 그러므로 해는 유일하다.)

### 📝 예제 3 (Example 3)

다음 시스템이 해를 가지는지 판별하라:

$$
\begin{alignedat}{4}
& & x_2 & - & 4x_3 & = & 8 \\
2x_1 & - & 3x_2 & + & 2x_3 & = & 1 \\
4x_1 & - & 8x_2 & + & 12x_3 & = & 1
\end{alignedat}
$$

**해법**

첨가 행렬은 다음과 같다.

$$
\begin{bmatrix}
0 & 1 & -4 & 8 \\
2 & -3 & 2 & 1 \\
4 & -8 & 12 & 1
\end{bmatrix}
$$

첫 번째 방정식에서 $x_1$을 얻기 위해, 1행과 2행을 교환한다:

$$
\begin{bmatrix}
2 & -3 & 2 & 1 \\
0 & 1 & -4 & 8 \\
4 & -8 & 12 & 1
\end{bmatrix}
$$

세 번째 방정식의 $4x_1$ 항을 제거하기 위해, 1행에 -2를 곱한 것을 3행에 더한다:

$$
\begin{bmatrix}
2 & -3 & 2 & 1 \\
0 & 1 & -4 & 8 \\
0 & -2 & 8 & -1
\end{bmatrix}
$$

다음으로, 두 번째 방정식의 $x_2$ 항을 사용하여 세 번째 방정식의 $-2x_2$ 항을 제거한다. 2행에 2를 곱한 것을 3행에 더한다:

$$
\begin{bmatrix}
2 & -3 & 2 & 1 \\
0 & 1 & -4 & 8 \\
0 & 0 & 0 & 15
\end{bmatrix}
$$

이제 첨가 행렬은 삼각 형태이다. 이를 올바르게 해석하기 위해 방정식 표기법으로 돌아가자:

$$
\begin{alignedat}{4}
2x_1 & - & 3x_2 & + & 2x_3 & = & 1 \\
& & x_2 & - & 4x_3 & = & 8 \\
& & & & 0 & = & 15
\end{alignedat}
$$

방정식 $0 = 15$는 $0x_1 + 0x_2 + 0x_3 = 15$의 축약형이다. 이 삼각 형태의 시스템은 명백히 내재된 모순을 가지고 있다. $0 = 15$라는 방정식은 절대 참이 될 수 없으므로, 위 방정식을 만족하는 $x_1, x_2, x_3$의 값은 존재하지 않는다. 원래 시스템과 이 시스템은 동일한 해집합을 가지므로, 원래 시스템은 해가 존재하지 않는다(inconsistent).

행렬 (7)의 마지막 행에 주목하라. 이 마지막 행은 삼각 형태의 해가 없는 시스템의 전형적인 형태이다.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
plt.rcParams['font.family'] ='Malgun Gothic'
plt.rcParams['axes.unicode_minus'] =False
# 3차원 그래프를 위한 설정
fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection='3d')

# x1, x2 값의 범위를 생성한다.
x1 = np.linspace(-5, 5, 10)
x2 = np.linspace(-5, 5, 10)
x1, x2 = np.meshgrid(x1, x2)

# 세 평면의 방정식을 정의한다.
# 1: x2 - 4x3 = 8  => x3 = (x2 - 8) / 4
x3_1 = (x2 - 8) / 4
# 2: 2x1 - 3x2 + 2x3 = 1 => x3 = (1 - 2x1 + 3x2) / 2
x3_2 = (1 - 2*x1 + 3*x2) / 2
# 3: 4x1 - 8x2 + 12x3 = 1 => x3 = (1 - 4x1 + 8x2) / 12
x3_3 = (1 - 4*x1 + 8*x2) / 12

# 각 평면을 그린다 (서로 평행하지 않지만 공통 교점이 없음)
ax.plot_surface(x1, x2, x3_1, alpha=0.6, rstride=100, cstride=100, color='r')
ax.plot_surface(x1, x2, x3_2, alpha=0.6, rstride=100, cstride=100, color='g')
ax.plot_surface(x1, x2, x3_3, alpha=0.6, rstride=100, cstride=100, color='b')

# 축 레이블을 설정한다.
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$x_3$')
ax.set_title('Inconsistent System: No Common Intersection Point')
ax.text2D(0.05, 0.95, "세 평면 모두에 놓이는 점이 없으므로 시스템은 해가 없다.", transform=ax.transAxes)

plt.show()
```

### ✅ 합리적인 답변 (Reasonable Answers)

방정식 시스템에 대한 하나 이상의 해를 구했다면, 찾은 해를 원래 방정식에 다시 대입하여 답을 확인하는 것을 기억하라. 예를 들어, 방정식 시스템

$$
\begin{alignedat}{4}
x_1 & - & 2x_2 & + & x_3 & = & 2 \\
x_1 & & & - & 2x_3 & = & -2 \\
& & x_2 & + & x_3 & = & 3
\end{alignedat}
$$

의 해로 $(2, 1, -1)$을 찾았다면, 이 해를 원래 방정식에 대입하여 다음을 얻을 수 있다.

$$2 - 2(1) + (-1) = -1 \neq 2$$
$$2 - 2(-1) = 4 \neq -2$$
$$1 + (-1) = 0 \neq 3$$

이제 원래 계산에 오류가 있었음이 분명해졌다. 산수를 다시 확인한 후 답이 $(2, 1, 2)$임을 발견했다면, 다음을 볼 수 있다.

$$2 - 2(1) + (2) = 2 = 2$$
$$2 - 2(2) = -2 = -2$$
$$1 + 2 = 3 = 3$$

이제 주어진 방정식 시스템에 대한 올바른 해를 구했다고 확신할 수 있다.

### 💻 수치적 참고사항 (Numerical Note)

실제 문제에서 선형방정식 시스템은 컴퓨터로 해결된다. 정사각 계수 행렬의 경우, 컴퓨터 프로그램은 거의 항상 여기서와 1.2절에서 주어진 소거 알고리즘을 사용하며, 정확도를 향상시키기 위해 약간 수정된다.

비즈니스와 산업 분야의 선형대수학 문제 대다수는 부동 소수점 산술을 사용하는 프로그램으로 해결된다. 숫자는 $\pm.d_{1}\dots d_{p} \times 10^{r}$ 형태의 소수로 표현되며, 여기서 $r$은 정수이고 소수점 오른쪽의 자릿수 $p$는 보통 8에서 16 사이이다. 이러한 숫자를 사용한 산술은 일반적으로 부정확한데, 결과가 저장된 자릿수로 반올림(또는 절사)되어야 하기 때문이다. "반올림 오차(Roundoff error)"는 $1/3$과 같은 숫자를 컴퓨터에 입력할 때도 발생하는데, 이는 십진수 표현이 유한한 수의 자릿수로 근사되어야 하기 때문이다. 다행히도, 부동 소수점 산술의 부정확성은 거의 문제를 일으키지 않는다. 이 책의 수치적 참고사항들은 나중에 경력에서 고려해야 할 수 있는 문제들에 대해 때때로 경고할 것이다.

### ✏️ 연습문제 풀이 (Solutions to Practice Problems)

1.  a. "손 계산"의 경우, 가장 좋은 선택은 3번 방정식과 4번 방정식을 교환하는 것이다. 또 다른 가능성은 3번 방정식을 $1/5$로 곱하는 것이다. 또는 3행의 $-1/5$배를 4행에 더하여 4행을 교체하는 것이다. (어떤 경우든, 2번 방정식의 $x_2$를 사용하여 1번 방정식의 $4x_2$를 제거하지 마라. 삼각 형태에 도달하고 $x_3$ 항과 $x_4$ 항이 첫 두 방정식에서 제거될 때까지 기다려라.)

    b. 시스템은 삼각 형태이다. 추가적인 단순화는 네 번째 방정식의 $x_4$에서 시작된다. $x_4$를 사용하여 그 위의 모든 $x_4$ 항을 제거하라. 이제 적절한 단계는 4번 방정식에 2를 곱한 것을 1번 방정식에 더하는 것이다. (그 후에 3번 방정식으로 이동하여 $1/2$을 곱한 다음, 그 방정식을 사용하여 그 위의 $x_3$ 항들을 제거하라.)

2.  첨가 행렬에 해당하는 시스템은 다음과 같다.

    $$
    \begin{alignedat}{4}
    x_1 &+& 5x_2 &+& 2x_3 &=& -6 \\
         & & 4x_2 &-& 7x_3 &=& 2 \\
         & &      & & 5x_3 &=& 0
    \end{alignedat}
    $$

    세 번째 방정식은 $x_3 = 0$을 만드는데, 이는 $x_3$에 대해 확실히 허용되는 값이다. 1번과 2번 방정식에서 $x_3$ 항을 제거한 후, 계속해서 $x_2$와 $x_1$에 대한 유일한 값을 풀 수 있다. 따라서 해가 존재하며, 그 해는 유일하다. 이 상황을 예제 3의 상황과 대조해보라.

3.  특정 숫자 목록이 해인지 확인하는 것은 쉽다. $x_1 = 3, x_2 = 4, x_3 = -2$로 설정하고 다음을 계산한다.
    $$5(3) - (4) + 2(-2) = 15 - 4 - 4 = 7$$
    $$-2(3) + 6(4) + 9(-2) = -6 + 24 - 18 = 0$$
    $$-7(3) + 5(4) - 3(-2) = -21 + 20 + 6 = 5$$
    첫 두 방정식은 만족되지만, 세 번째는 만족되지 않으므로 $(3, 4, -2)$는 시스템의 해가 아니다. 대입할 때 괄호를 사용하는 것에 주목하라. 산술 오류를 방지하기 위해 강력히 권장된다.

4.  두 번째 방정식을 첫 번째 방정식의 3배를 더한 것으로 교체하면, 시스템은 다음과 같이 된다.
    $$2x_1 - x_2 = h$$
    $$0 = k + 3h$$
    만약 $k + 3h$가 0이 아니면, 시스템은 해가 없다. 시스템은 $k + 3h = 0$을 만족하는 $h$와 $k$의 모든 값에 대해 해를 가진다.
